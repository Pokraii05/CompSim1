## Problem 1: Statement

Calculate the exact solution of the one-dimensional Ising model with $N$ spins for free boundary conditions:

a) Sketch the calculation of the partition function $Z$ by successive summation of spins.

b) Calculate the internal energy $U = \langle E\rangle = -\dfrac{\partial \ln Z}{\partial \beta}$ and the specific heat $C = \dfrac{dU}{dT}$.

c) How do $u = U/N$ and $c = C/N$ behave in the thermodynamic limit $N \to \infty$?

## Problem 1: Solution


(a)

Let us consider a one dimensional Ising model with $N$ spins for free boundary conditions. We can write the energy of such system as follows

$$
E=-J \sum_{i=1}^{N-1}s_i s_{i+1} \tag{1}
$$

where the sum is over all pairs of neighbouring spins $i$ and $i+1$ and $J$ is an exchange constant assumed to be positive. For algebraic simplicity's sake, we can expand eq. (1) as follows

$$
E=-J(s_1s_2+s_2s_3+\cdots+s_{N-2}s_{N-1}+s_{N-1}s_N) \tag{2}
$$

Furthermore, by definition, we know that the partition function is given by the following equation:

$$
Z=\sum_{\{s_i=\pm1\}} e^{\frac{-E}{k_B T}} \tag{3}
$$

Plugging in eq. (2) into eq. (3) we arrive at eq. (4)

$$
Z=\sum_{\{s_i=\pm1\}} e^{\frac{J}{k_B T}(s_1s_2+s_2s_3+\cdots+s_{N-2}s_{N-1}+s_{N-1}s_N)} \tag{4}
$$

This can be expanded over all states as follows:

$$
Z=\sum_{s_1=\pm1}\cdots\sum_{s_N=\pm1}e^{\frac{J}{k_B T}(s_1s_2+s_2s_3+\cdots+s_{N-2}s_{N-1}+s_{N-1}s_N)} \tag{5}
$$

Notice how the term $s_N$ appears only in the last term. Therefore, we can extract the sum accounting for the last term as shown in eq. (6)

$$
Z=\sum_{s_1=\pm1}\cdots\sum_{s_{N-1}=\pm1}e^{\frac{J}{k_B T}(s_1s_2+s_2s_3+\cdots+s_{N-2}s_{N-1})} \sum_{s_N=\pm1} e^{\frac{J}{k_B T}s_{N-1}s_N} \tag{6}
$$

Since $s_N$ can only be $+1$ or $-1$, we calculate the last term as:

$$
\sum_{s_N=\pm1} e^{\frac{J}{k_B T}s_{N-1}s_N} = e^{\frac{J}{k_B T}s_{N-1}} + e^{-\frac{J}{k_B T}s_{N-1}} \tag{7}
$$

Recalling the hyperbolic cosine definition given by eq. (8)

$$
\cosh(\theta)=\frac{e^{\theta}+e^{-\theta}}{2} \tag{8}
$$

One can rewrite eq. (7) in terms of the hyperbolic cosine. Since we know that $s_{N-1}=\pm1$ and since the hyperbolic cosine is an even function we can further simplify the expression as follows

$$
\sum_{s_N=\pm1} e^{\frac{J}{k_B T}s_{N-1}s_N} = 2\cosh\left(\frac{Js_{N-1}}{k_B T}\right) = 2\cosh\left(\frac{J}{k_B T}\right) \tag{9}
$$

Plugging this back into eq. (6)

$$
Z=\sum_{s_1=\pm1}\cdots\sum_{s_{N-1}=\pm1}e^{\frac{J}{k_B T}(s_1s_2+s_2s_3+\cdots+s_{N-2}s_{N-1})}\, 2\cosh\left(\frac{J}{k_B T}\right) \tag{10}
$$

we realize this can be done recursively from the "back" of the expression towards the front, i.e.:

$$
Z=\sum_{s_1=\pm1}\cdots\sum_{s_{N-2}=\pm1}e^{\frac{J}{k_B T}(s_1s_2+s_2s_3+\cdots+s_{N-3}s_{N-2})}\, 2\cosh\left(\frac{J}{k_B T}\right) 2\cosh\left(\frac{J}{k_B T}\right) \tag{11}
$$

$$
Z=\sum_{s_1=\pm1}\cdots\sum_{s_{N-3}=\pm1}e^{\frac{J}{k_B T}(s_1s_2+s_2s_3+\cdots+s_{N-4}s_{N-3})}\, 2\cosh\left(\frac{J}{k_B T}\right) 2\cosh\left(\frac{J}{k_B T}\right) 2\cosh\left(\frac{J}{k_B T}\right) \tag{12}
$$

and so on, $N-1$ times, until you arrive at the following

$$
Z=\left[2\cosh\left(\frac{J}{k_B T}\right)\right]^{N-1}\sum_{s_1=\pm1}1 \tag{13}
$$

and since $\sum_{s_1=\pm1}1=1+1=2$, finally we get:

$$
\boxed{Z=2^N\left[\cosh\left(\frac{J}{k_B T}\right)\right]^{N-1}} \tag{14}
$$

(b)

Now we are tasked with calculating the internal energy $U=\langle E\rangle=-\frac{\partial \ln Z}{\partial \beta}$ and the specific heat $C=\frac{dU}{dT}$.

First we calculate the internal energy by taking a partial derivative with respect to $\beta$, which is assumed to be equal to $\beta=\frac{1}{k_B T}$.

$$
\begin{aligned}
U &= -\frac{\partial \ln Z}{\partial \beta} = -\frac{\partial}{\partial \beta} \ln\left(2^N\left[\cosh(J\beta)\right]^{N-1}\right) \\
&= -\frac{\partial}{\partial \beta} \left(N\ln 2 + (N-1)\ln\left(\cosh(J\beta)\right)\right) \\
&= -\frac{\partial}{\partial \beta} (N-1) \ln\left(\cosh(J\beta)\right)
\end{aligned}
$$

by employing the chain rule we get

$$
= -(N-1)\frac{1}{\cosh(J\beta)}\sinh(J\beta)J
$$

furthermore using the definition of the hyperbolic tangent we conclude

$$
\boxed{U = -J(N-1)\tanh(J\beta)} \tag{15}
$$

or equivalently

$$
\boxed{U = -J(N-1)\tanh\left(\frac{J}{k_B T}\right)} \tag{16}
$$

If we wish to calculate the specific heat, we take a derivative of the internal energy function with respect to temperature.

$$
\begin{aligned}
C &= \frac{dU}{dT} = \frac{d}{dT}\left(-J(N-1)\tanh\left(\frac{J}{k_B T}\right)\right) \\
&= -J(N-1)\frac{d}{dT}\tanh\left(\frac{J}{k_B T}\right)
\end{aligned}
$$

by knowing that $\frac{d \tanh(\theta)}{d\theta}=\operatorname{sech}^2(\theta)$ we conclude

$$
\boxed{C = \frac{J^2}{k_B T^2}(N-1)\operatorname{sech}^2\left(\frac{J}{k_B T}\right)} \tag{17}
$$

(c)

Finally, we observe how $u=U/N$ and $c=C/N$ behave in the thermodynamic limit $N \to \infty$

$$
u=\frac{-J(N-1)\tanh\left(\frac{J}{k_B T}\right)}{N}
$$

$$
c=\frac{\frac{J^2}{k_B T^2}(N-1)\operatorname{sech}^2\left(\frac{J}{k_B T}\right)}{N}
$$

taking the limit as $N \to \infty$ for both expressions

$$
\begin{aligned}
&\lim_{N \to \infty} \frac{-J(N-1)\tanh\left(\frac{J}{k_B T}\right)}{N} \\
&= -J\tanh\left(\frac{J}{k_B T}\right) \lim_{N \to \infty} \frac{N-1}{N} \\
&= -J\tanh\left(\frac{J}{k_B T}\right) \lim_{N \to \infty} \left(1-\frac{1}{N}\right) \\
&\lim_{N \to \infty} u= \boxed{-J\tanh\left(\frac{J}{k_B T}\right)}
\end{aligned} \tag{18}
$$

$$
\begin{aligned}
&\lim_{N \to \infty} \frac{\frac{J^2}{k_B T^2}(N-1)\operatorname{sech}^2\left(\frac{J}{k_B T}\right)}{N} \\
&= \lim_{N \to \infty} \frac{J^2}{k_B T^2}\operatorname{sech}^2\left(\frac{J}{k_B T}\right)\frac{N-1}{N} \\
&= \frac{J^2}{k_B T^2}\operatorname{sech}^2\left(\frac{J}{k_B T}\right) \lim_{N \to \infty} \frac{N-1}{N} \\
&= \frac{J^2}{k_B T^2}\operatorname{sech}^2\left(\frac{J}{k_B T}\right) \lim_{N \to \infty} \left(1-\frac{1}{N}\right) \\
& \boxed{\lim_{N \to \infty} c=\frac{J^2}{k_B T^2}\operatorname{sech}^2\left(\frac{J}{k_B T}\right)}
\end{aligned} \tag{19}
$$

## Problem 2: Statement

a) Plot $u$ and $c$ of exercise 1 as function of temperature $T$ (e.g., with "gnuplot") for various finite $N$ ($= 20, 40, \dots, 100$) and discuss the curves (for example: Does a maximum exist for $c$ and, if yes, at which temperature?).

b) Study how the thermodynamic limit $N \to \infty$ is approached?

## Problem 2: Solution

(a)

We simplify expressions given by eq. (16) and (17)

$$
u(N,T)=-J\frac{(N-1)}{N}\tanh\left(\frac{J}{k_B T}\right) \tag{20}
$$

$$
c(N,T)=\frac{J^2}{k_B T^2}\frac{(N-1)}{N}\operatorname{sech}^2\left(\frac{J}{k_B T}\right) \tag{21}
$$

Then we plot $u$ as a function of $T$ for several values of $N$, as demanded in the problem statement.

```python
import numpy as np
from matplotlib import pyplot as plt

J = 1.0
k_b = 1.0

N_values = [20, 40, 60, 80, 100]

T = np.linspace(0.01, 50, 500)

def u(N,T):
  return -J * (N-1) / N * np.tanh(J/(k_b*T))

for N in N_values:
    plt.plot(T,u(N,T), label=f"N={N}")

plt.xlabel("T")
plt.ylabel("u(N,T)")
plt.legend()
plt.show()
```

Figure 1. Dependence of $u(N,T)$ on $T$, for $N=20,40,60,80,100$

As we can see on Fig. 1, as $T \to \infty$,  $u(N,T) \to 0$. Additionally we see that as N is increased, the curves start off at lower and lower $u(N,T)$. More mathematically speaking as $T \to 0$, $u(N,T) \to -J(N-1)/N$.



Furthermore we plot $c$ as a function of $T$ for several values of $N$, as stated in the problem statement.

```python
def c(N, T):
    return J**2/(k_b*T**2) * (N-1)/N * 1/(np.cosh(J/(k_b*T))**2)

for N in N_values:
    plt.plot(T,c(N,T), label=f"N={N}")

plt.xlabel("T")
plt.ylabel("c(N,T)")
plt.legend()
plt.show()
```

Figure 2. Dependence of $c(N,T)$ on $T$, for $N=20,40,60,80,100$

As we can see on Fig. 2, as $T \to \infty$, $c(N,T) \to 0$. Additionally, there exists a clear maximum of $c(N,T)$. $c(N,T)$ increases from zero, reaches a maximum near $T=1$, and then decreases towards zero. 

Let us plot a plot with a smaller $T$ bandwidth to identify at which exact $T$ value the maximum of $c(N,T)$ appears . We use `np.argmax` to find the largest values of $c$ for each $N$, and then find the corresponding $T$ at which it appears.

```python
T_2 = np.linspace(0.01, 5, 500)

for N in N_values:
    c_val = c(N, T_2)
    plt.plot(T_2, c_val, label=f"N={N}")

    i_max = np.argmax(c_val)
    T_max = T_2[i_max]

    print(f"N={N}: T_max={T_max:.4f}")

plt.axvline(0.83, linestyle="--", color="red", label="T=0.83")
plt.xlabel("T")
plt.ylabel("c(N,T)")
plt.legend()
plt.show()
```

Figure 3. Dependence of $c(N,T)$ on $T$, for $N=20,40,60,80,100$ and $T\le 5$

On Fig. 3 we can much more clearly see that the maximum of $c(N,T)$ exists at $T$ around $1$. Numerically we determine this number to be $T=0.83$. We also notice that increasing $N$ changes the height of the curves but not the temperature at which the heat capacity reaches its maximum.

(b)

Now we wish to study how the thermodynamic limit of $N \to \infty$ behaves for both $u(N,T)$ and $c(N,T)$.

Let us remember eq. (18) and (19) derived in the previous task.

$$\lim_{N \to \infty} u= -J\tanh\left(\frac{J}{k_B T}\right)$$

$$\lim_{N \to \infty} c=\frac{J^2}{k_B T^2}\operatorname{sech}^2\left(\frac{J}{k_B T}\right)$$

It is useful to notice that $u$ and $c$ themselves given by eq. (20) and (21) are nothing more than

$$u(N,T)=\frac{N-1}{N} \lim_{N \to \infty} u$$

$$c(N,T)=\frac{N-1}{N} \lim_{N \to \infty} c$$

Since the limits in eq. (18) and (19) are independent of $N$, now we see that the thermodynamic limit is approached through the factor $\frac{N-1}{N}=1-\frac{1}{N}$. This means that the error from the thermodynamic limit decreases proportionally with $\frac{1}{N}$.

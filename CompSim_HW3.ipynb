In this task however, we are asked to determine the density of states $\Omega (E,M)$. From the lecture and previous homeworks we know that the partition function for the 1D Ising model with free boundary conditions is given by eq. X.

$$ Z(\beta)=\sum_{\{s\}} e^{-\beta E(\{s\})}$$

Furthermore, we know that eq. X can also be rewritten as 

$$ Z(\beta)=\sum_{E} \Omega(E) e^{-\beta E}$$

where $\Omega(E)$ is the density of states. In the current exact enumeration code we calculate E for every spin configuration, and in order to determine $\Omega(E)$ we can just save every E in a dictionary using a conditional statement: If E is not in dictionary, add E to dictionary with key 1, else if E is in dictionary, add 1 to the already existing key for E.

Since we know that there are 20 spins, and each of them can take one of two values, the total amount of possible combinations of the 20 spins is $2^20$. This means that 

$$\sum_{E} \Omega(E) = 2^20 = 1048576$$

Using this we can check if the determined $\Omega(E)$ is correct.


```python
import math

def exp(x):
    return math.exp(x)

spins=[-1,1]

T=1.0
J=1.0
k_b=1.0

print("EXACT ENUMERATION RESULTS:")

Z_20=0
E_20=0
distr={}
for s1 in spins:
    for s2 in spins:
        for s3 in spins:
            for s4 in spins:
                for s5 in spins:
                    for s6 in spins:
                        for s7 in spins:
                            for s8 in spins:
                                for s9 in spins:
                                    for s10 in spins:
                                        for s11 in spins:
                                            for s12 in spins:
                                                for s13 in spins:
                                                    for s14 in spins:
                                                        for s15 in spins:
                                                            for s16 in spins:
                                                                for s17 in spins:
                                                                    for s18 in spins:
                                                                        for s19 in spins:
                                                                            for s20 in spins:
                                                                                E_20=-J/(k_b*T)*(s1*s2+s2*s3+s3*s4+s4*s5+s5*s6+s6*s7+s7*s8+s8*s9+s9*s10+s10*s11+s11*s12+s12*s13+s13*s14+s14*s15+s15*s16+s16*s17+s17*s18+s18*s19+s19*s20)
                                                                                Z_20+=exp(-E_20)
                                                                                if E_20 in distr:
                                                                                    distr[E_20]+=1
                                                                                else:
                                                                                    distr.update({E_20 : 1})
                                                                                

print("For N=20, Z =", Z_20)
print("OMEGA(E):")
print(distr)
print("SUM_E OMEGA(E):")
print(sum(distr.values()))
if sum(distr.values()) == 1048576:
    print("-> check passed")
```

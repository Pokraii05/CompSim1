## Problem 5: Solution

First let us look at the code written for Problem 3. We wrote this code in order to determine the partition function Z. 

```python
import math

def exp(x):
    return math.exp(x)

spins=[-1,1]

T=1.0
J=1.0
k_b=1.0

print("EXACT ENUMERATION RESULTS:")

Z_20=0
E_20=0
distr={}
for s1 in spins:
    for s2 in spins:
        for s3 in spins:
            for s4 in spins:
                for s5 in spins:
                    for s6 in spins:
                        for s7 in spins:
                            for s8 in spins:
                                for s9 in spins:
                                    for s10 in spins:
                                        for s11 in spins:
                                            for s12 in spins:
                                                for s13 in spins:
                                                    for s14 in spins:
                                                        for s15 in spins:
                                                            for s16 in spins:
                                                                for s17 in spins:
                                                                    for s18 in spins:
                                                                        for s19 in spins:
                                                                            for s20 in spins:
                                                                                E_20=-J/(k_b*T)*(s1*s2+s2*s3+s3*s4+s4*s5+s5*s6+s6*s7+s7*s8+s8*s9+s9*s10+s10*s11+s11*s12+s12*s13+s13*s14+s14*s15+s15*s16+s16*s17+s17*s18+s18*s19+s19*s20)
                                                                                Z_20+=exp(-E_20)
                                                                                

print("For N=20, Z =", Z_20)
```

In this task however, we are asked to determine the density of states $\Omega (E,M)$. From the lecture and previous homeworks we know that the partition function for the 1D Ising model with free boundary conditions is given by eq. (1).

$$
Z(\beta)=\sum_{\{s\}} e^{-\beta E(\{s\})} \tag{1}
$$

Furthermore, we know that eq. (1) can also be rewritten as 

$$
Z(\beta)=\sum_{E} \Omega(E) e^{-\beta E} \tag{2}
$$

where $\Omega(E)$ is the density of states. In the current exact enumeration code we calculate E for every spin configuration, and in order to determine $\Omega(E)$ we can just save every E in a dictionary using a conditional statement: If E is not in dictionary, add E to dictionary with key 1, else if E is in dictionary, add 1 to the already existing key for E.

Since we know that there are 20 spins, and each of them can take one of two values, the total amount of possible combinations of the 20 spins is $2^{20}$. This means that 

$$
\sum_{E} \Omega(E) = 2^{20} = 1048576 \tag{3}
$$

Using this we can check if the determined $\Omega(E)$ is correct.


```python
import math

def exp(x):
    return math.exp(x)

spins=[-1,1]

J=1.0

print("EXACT ENUMERATION RESULTS:")

Z_20=0
E_20=0
distr={}
for s1 in spins:
    for s2 in spins:
        for s3 in spins:
            for s4 in spins:
                for s5 in spins:
                    for s6 in spins:
                        for s7 in spins:
                            for s8 in spins:
                                for s9 in spins:
                                    for s10 in spins:
                                        for s11 in spins:
                                            for s12 in spins:
                                                for s13 in spins:
                                                    for s14 in spins:
                                                        for s15 in spins:
                                                            for s16 in spins:
                                                                for s17 in spins:
                                                                    for s18 in spins:
                                                                        for s19 in spins:
                                                                            for s20 in spins:
                                                                                E_20=-J*(s1*s2+s2*s3+s3*s4+s4*s5+s5*s6+s6*s7+s7*s8+s8*s9+s9*s10+s10*s11+s11*s12+s12*s13+s13*s14+s14*s15+s15*s16+s16*s17+s17*s18+s18*s19+s19*s20)
                                                                                Z_20+=exp(-E_20)
                                                                                if E_20 in distr:
                                                                                    distr[E_20]+=1
                                                                                else:
                                                                                    distr.update({E_20 : 1})
                                                                                

print("For N=20, Z =", Z_20)
print("OMEGA(E):")
for E, omega in distr.items():
    print(f"E={E}: {omega}")
print("SUM_E OMEGA(E) =", sum(distr.values()))
if sum(distr.values()) == 1048576:
    print("-> check passed")
```

(a)

Furthermore, we are asked to find $\langle E \rangle(T)$ and the specific heat $C(T)$ for $h=0$ using our predetermined $\Omega (E,M)$

From Homework 1 we know that $\langle E\rangle = -\dfrac{\partial \ln Z}{\partial \beta}$ and the specific heat is $C = \dfrac{d \langle E \rangle}{dT}$.


If we use our formulation of the partition function Z given by eq. (2) we can rewrite $\langle E \rangle$ as follows:

$$
\begin{aligned}
\langle E\rangle &= -\dfrac{\partial \ln Z}{\partial \beta} \\
&= \frac{-1}{\sum_{E} \Omega(E) e^{-\beta E}} \frac{\partial }{\partial \beta} \sum_{E} \Omega(E) e^{-\beta E} = \frac{\sum_{E} E\Omega(E) e^{-\beta E}}{\sum_{E} \Omega(E) e^{-\beta E}}
\end{aligned} \tag{4}
$$


Then $\langle E \rangle(T)$ can be computed as follows:

```python
Z_d = 0
k_b = 1.0
T = 1.0
E_exp = 0

E_top = 0
E_bottom=0

for E in distr:
    E_top+=E*distr[E]*exp(-E/(k_b*T))
    E_bottom+=distr[E]*exp(-E/(k_b*T))

E_exp=E_top/E_bottom

for E in distr:
    Z_d+=distr[E]*exp(-E/(k_b*T))

print("For N=20, Z =", Z_d)
print("For N=20, <E> =", E_exp)
```

Furthermore we can derive the specific heat as follows:

$$
C=\frac{d \langle E \rangle}{dT}=\frac{d}{dT}\frac{\sum_{E} E\Omega(E) e^{-\frac{1}{k_bT} E}}{\sum_{E} \Omega(E) e^{-\frac{1}{k_bT} E}}
=\frac{1}{k_B T^2}
\left[
\frac{\displaystyle\sum_E E^2\Omega(E)e^{-\frac{E}{k_B T}}}
{\displaystyle\sum_E \Omega(E)e^{-\frac{E}{k_B T}}}
-
\left(
\frac{\displaystyle\sum_E E\Omega(E)e^{-\frac{E}{k_B T}}}
{\displaystyle\sum_E \Omega(E)e^{-\frac{E}{k_B T}}}
\right)^2
\right] \tag{5}
$$



```python
E_1 = 0
E_2 = 0
E_3 = 0
E_4 = 0

for E in distr:
    E_1+=E*E*distr[E]*exp(-E/(k_b*T))
    E_2+=distr[E]*exp(-E/(k_b*T))
    E_3+=E*distr[E]*exp(-E/(k_b*T))
    E_4=E_2

C=1/(k_b*T*T)*(E_1/E_2-(E_3/E_4)**2)
print("For N=20, C =", C)
```

Finally we plot both $\langle E \rangle (T)$ and $C(T)$ on the interval $T \in (0,2]$

```python
import numpy as np
from matplotlib import pyplot as plt

J = 1.0
k_b = 1.0

T = np.linspace(0.1, 2, 100)

def E(T):
    E_top = 0
    E_bottom = 0

    for E in distr:
        E_top += E*distr[E]*exp(-E/(k_b*T))
        E_bottom += distr[E]*exp(-E/(k_b*T))

    return E_top/E_bottom

def C(T):
    E_1 = 0
    E_2 = 0
    E_3 = 0
    E_4 = 0

    for E in distr:
        E_1 += E*E*distr[E]*exp(-E/(k_b*T))
        E_2 += distr[E]*exp(-E/(k_b*T))
        E_3 += E*distr[E]*exp(-E/(k_b*T))
        E_4 = E_2

    return 1/(k_b*T*T)*(E_1/E_2-(E_3/E_4)**2)

plt.plot(T, [E(t) for t in T], label=r"$\langle E \rangle (T)$")
plt.plot(T, [C(t) for t in T], label="C(T)")

plt.xlabel("T")
plt.legend()
plt.show()
```

Figure 1. Dependence of $\langle E \rangle (T)$ and $C(T)$ on $T$, for $N=20$ and $h=0$

Now we wish to determine the $\langle E \rangle (T)$ and $\langle M \rangle(T)$ by means of $\Omega (E,M)$ for $h \neq 0$. This means that we need to expand our definition of $\Omega (E)$ into $\Omega(E,M)$. As discussed in Homework 2, we can extend the 1D Ising model with free boundary conditions to cases with magnetization by adding another term into the energy function as shown by eq. (6)

$$
E=-J \sum_{i=1}^{N-1}s_i s_{i+1} -h\sum_{i=1}^{N}s_i \tag{6}
$$


Then the code for the exact enumeration would be the following, where for h=0 we just get the case without magnetization:

```python
import math

def exp(x):
    return math.exp(x)

spins=[-1,1]

T=1.0
J=1.0
k_b=1.0
h=0.0

print("EXACT ENUMERATION RESULTS:")

Z_20=0
E_20=0
distr={}
for s1 in spins:
    for s2 in spins:
        for s3 in spins:
            for s4 in spins:
                for s5 in spins:
                    for s6 in spins:
                        for s7 in spins:
                            for s8 in spins:
                                for s9 in spins:
                                    for s10 in spins:
                                        for s11 in spins:
                                            for s12 in spins:
                                                for s13 in spins:
                                                    for s14 in spins:
                                                        for s15 in spins:
                                                            for s16 in spins:
                                                                for s17 in spins:
                                                                    for s18 in spins:
                                                                        for s19 in spins:
                                                                            for s20 in spins:
                                                                                E_20=-J*(s1*s2+s2*s3+s3*s4+s4*s5+s5*s6+s6*s7+s7*s8+s8*s9+s9*s10+s10*s11+s11*s12+s12*s13+s13*s14+s14*s15+s15*s16+s16*s17+s17*s18+s18*s19+s19*s20)
                                                                                M=(s1+s2+s3+s4+s5+s6+s7+s8+s9+s10+s11+s12+s13+s14+s15+s16+s17+s18+s19+s20)
                                                                                Z_20+=exp(-(E_20-h*M)/(k_b*T))
                                                                                

print("For N=20, Z =", Z_20)
```

Then the density of states $\Omega (E,M)$ can be determined as follows:

```python
import math

def exp(x):
    return math.exp(x)

spins=[-1,1]

T=1.0
J=1.0
k_b=1.0
h=0.0

print("EXACT ENUMERATION RESULTS:")

Z_20=0
E_20=0
distr={}
for s1 in spins:
    for s2 in spins:
        for s3 in spins:
            for s4 in spins:
                for s5 in spins:
                    for s6 in spins:
                        for s7 in spins:
                            for s8 in spins:
                                for s9 in spins:
                                    for s10 in spins:
                                        for s11 in spins:
                                            for s12 in spins:
                                                for s13 in spins:
                                                    for s14 in spins:
                                                        for s15 in spins:
                                                            for s16 in spins:
                                                                for s17 in spins:
                                                                    for s18 in spins:
                                                                        for s19 in spins:
                                                                            for s20 in spins:
                                                                                E_20=-J*(s1*s2+s2*s3+s3*s4+s4*s5+s5*s6+s6*s7+s7*s8+s8*s9+s9*s10+s10*s11+s11*s12+s12*s13+s13*s14+s14*s15+s15*s16+s16*s17+s17*s18+s18*s19+s19*s20)
                                                                                M_20=(s1+s2+s3+s4+s5+s6+s7+s8+s9+s10+s11+s12+s13+s14+s15+s16+s17+s18+s19+s20)
                                                                                Z_20+=exp(-(E_20-h*M_20)/(k_b*T))

                                                                                if (E_20, M_20) in distr:
                                                                                    distr[(E_20, M_20)]+=1
                                                                                else:
                                                                                    distr.update({(E_20, M_20) : 1})
                                                                                

print("For N=20, Z =", Z_20)
print("OMEGA(E,M):")
for (E, M), omega in distr.items():
    print(f"E={E}, M={M}: {omega}")
print("SUM_E,M OMEGA(E,M) =", sum(distr.values()))
if sum(distr.values()) == 1048576:
    print("-> check passed")
```

We can find $\langle E \rangle(T)$ and $\langle M \rangle(T)$ similarly as in part (a). The partition function is

$$
Z(T,h)=\sum_{E,M}\Omega(E,M)e^{-\beta(E-hM)} \tag{7}
$$

Then $\langle E \rangle(T)$ and $\langle M \rangle(T)$ are as follows:

$$
\langle E\rangle(T)
=
\frac{
\displaystyle\sum_{E,M}E\Omega(E,M)e^{-\beta(E-hM)}
}{
\displaystyle\sum_{E,M}\Omega(E,M)e^{-\beta(E-hM)}
} \tag{8}
$$

$$
\langle M\rangle(T)
=
\frac{1}{\beta}\frac{\partial\ln Z}{\partial h}
=
\frac{
\displaystyle\sum_{E,M}M\Omega(E,M)e^{-\beta(E-hM)}
}{
\displaystyle\sum_{E,M}\Omega(E,M)e^{-\beta(E-hM)}
} \tag{9}
$$


Finally, we can plot both $\langle E \rangle$ and $\langle M \rangle$ as a function of temperature $T$ on interval $T \in (0, 2]$ for $h = 0.5$.

```python
import numpy as np
from matplotlib import pyplot as plt

J = 1.0
k_b = 1.0

T = np.linspace(0.1, 2, 100)

def E(T):
    E_top = 0
    E_bottom = 0
    beta = 1.0 / (k_b * T)
    h = 0.5

    for E, M in distr:
        E_top += E * distr[E, M] * exp(-beta*(E - h * M))
        E_bottom += distr[E, M] * exp(-beta*(E - h * M))

    return E_top / E_bottom

def M(T):
    M_top = 0
    M_bottom = 0
    beta = 1.0 / (k_b * T)
    h = 0.5

    for E, M in distr:
        M_top += M * distr[E, M] * exp(-beta*(E - h * M))
        M_bottom += distr[E, M] * exp(-beta*(E - h * M))

    return M_top / M_bottom

plt.plot(T, [E(t) for t in T], label=r"$\langle E \rangle (T)$")
plt.plot(T, [M(t) for t in T], label=r"$\langle M \rangle (T)$")

plt.xlabel("T")
plt.legend()
plt.show()
```

Figure 2. Dependence of $\langle E \rangle (T)$ and $\langle M \rangle (T)$ on $T$, for $N=20$ and $h=0.5$
